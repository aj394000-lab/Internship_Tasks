## Q1. Find the Frequency of Each Word in a Sentence.

```python
a = 'python is easy and python is powerful'
b = {}
a = a.split()
for i in a:
    if i in b:
        b[i] += 1
    else:
        b[i] = 1
print(b)
```

## Q3. Find the First Non-Repeating Character in a Word.

```python
a = 'swiss'
b = {}
for i in a:
    if i in b:
        b[i] += 1
    else:
        b[i] = 1
d = 0
for i in a:
    if b[i] == 1:
        d = i
        break
print(d)
```

## Q5. Find the First Repeating Character in a String.

```python
a = 'programming'
b = {}
for i in a:
    if i in b:
        b[i] += 1
    else:
        b[i] = 1
d = 0
for i in a:
    if b[i] > 1:
        d = i
        break
print(d)
```

## Q6. Move All Zeros to the End of a List.

```python
a = [0, 3, 0, 1, 2, 0, 4]
d = []
for i in a:
    if i > 0:
        d.append(i)
for i in a:
    if i == 0:
        d.append(i)
print(d)
```

## Q7. Find the Second Highest Distinct Number in a List.

```python
a = [10, 20, 40, 30, 40, 20]
d = sorted(list(set(a)), reverse=True)
print(d[1])
```

## Q8. Remove Duplicates Without Changing the Original Order.

```python
a = [4, 2, 4, 5, 2, 7, 5]
seen = set()
result = []

for i in a:
    if i not in seen:
        seen.add(i)
        result.append(i)

a = result
print(a)
```

## Q9. Find the Missing Number from 1 to N.

```python
nums = [1, 2, 4, 5, 6]
N = 6
a = N * (N + 1) // 2
f = sum(nums)
d = a - f
print(d)
```

## Q10. Find Two Numbers Whose Sum Equals a Target.

```python
nums = [2, 7, 11, 15]
target = 9
result = None
d = set()
for i in nums:
    a = target - i
    if a in d:
        result = [a, i]
    d.add(i)

print(result)
```

## Q11. Find the Number Closest to Zero.

```python
a = [-8, -3, 2, 5, -2]
d = min(a, key=abs)
print(d)
```

## Q12. Find Common Elements Between Two Lists.

```python
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]
d = [i for i in list1 if i in list2]
print(d)
```

## Q13. Rotate a List to the Right by K Positions.

```python
from collections import deque
nums = [1, 2, 3, 4, 5]
K = 2
d = deque(nums)
d.rotate(K)
nums = list(d)
print(nums)
```

## Q14. Find the Maximum Sum of Two Consecutive Elements.

```python
a = [4, 7, 2, 9, 5]
max_sum = a[0] + a[1]
for i in range(len(a) - 1):
    current_sum = a[i] + a[i + 1]
    if current_sum > max_sum:
        max_sum = current_sum

print(max_sum)
```

## Q15. Find All Pairs of Numbers Whose Sum Equals a Target.

```python
nums = [2, 4, 3, 5, 7, 8, 1]
target = 9
result = [[nums[i], nums[j]] for i in range(len(nums)) for j in range(i + 1, len(nums)) if nums[i] + nums[j] == target]
print(result)
```
