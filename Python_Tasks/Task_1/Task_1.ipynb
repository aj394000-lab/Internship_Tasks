```python
# Q35
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
# copy.copy perform a shallow copy.
# Inner lists are still referenced by both.
```

```python
# Q36
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
# copy.deepcopy make a duplicates copy.
# so modifying the inner elements of b does not affect a.
```

```python
# Q37
import copy

a = [[10], [20]]
b = a.copy()
a[1].append(30)
print(b)
# a.copy() is a shallow copy. The nested lists are still sharing, so changing in a  is also ffects b.
```

```python
# Q38
import copy

a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)
# Using deepcopy completely decouples all nested structures, so changes inside 'a' have no effect on 'b'.
```

```python
# Q39
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
print(a is b)
print(a[0] is b[0])
# The outer list is copied (False), but the internal objects are not copied (True) due to the shallow nature of copy().
```

```python
# Q44
def test(*args):
    print(args)

test(1, 2)
# The *args parameter packs all positional arguments passed to the function into a tuple.
```

```python
# Q45
def test(*args):
    print(type(args))

test("a", "b", "c")
# The positional arguments captured by *args are always packed into a tuple object.
```

```python
# Q46
def test(**kwargs):
    print(kwargs)

test(name="Ram", age=25)
# The **kwargs parameter captures keyword arguments and packs them into a dictionary.
```

```python
# Q47
def test(**kwargs):
    print(type(kwargs))

test(a=10, b=20)
# Keyword arguments captured by **kwargs are always packaged in a dictionary.
```
